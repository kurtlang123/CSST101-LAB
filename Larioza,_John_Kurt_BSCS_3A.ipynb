```r
# Numerical variables
x <- 10.5
y <- 3

# Basic arithmetic operations
add <- x + y   # Addition
sub <- x - y   # Subtraction
mul <- x * y   # Multiplication
div <- x / y   # Division
mod <- x %% y  # Modulus
exp <- x ^ y   # Exponentiation

print(add)
print(sub)
print(mul)
print(div)
print(mod)
print(exp)
```

```r
user_name <- readline(prompt = "Enter your name: ")

# Prompting for a numeric input (converted from string to numeric)
user_age <- as.numeric(readline(prompt = "Enter your age: "))

# Displaying the input back to the user
print(paste("Hello,", user_name, "! You are", user_age, "years old."))

# Performing a simple calculation with the input
years_to_hundred <- 100 - user_age
cat(user_name, "will turn 100 in", years_to_hundred, "years!\n")
```

```r
score <- 75

# Basic if...else statement
if (score >= 75) {
  print("Status: Passed")
} else {
  print("Status: Failed")
}

# Multiple conditions using if...else if...else
if (score >= 90) {
  grade <- "A"
} else if (score >= 80) {
  grade <- "B"
} else if (score >= 70) {
  grade <- "C"
} else {
  grade <- "F"
}

cat("Grade:", grade, "\n")

# Vectorized condition using ifelse()
# Great for quick checks or applying logic across vectors!
status <- ifelse(score >= 75, "High Distinction", "Standard Pass")
print(status)
```
