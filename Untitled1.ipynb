```r
classify_grade <- function(grade) {
  if (grade >= 90 && grade <= 100) {
    return("EXCELLENT")
  } else if (grade >= 85 && grade <= 89) {
    return("VERY GOOD")
  } else if (grade >= 80 && grade <= 84) {
    return("GOOD")
  } else if (grade >= 75 && grade <= 79) {
    return("PASSED")
  } else {
    return("FAILED")
  }
}

# Example usage:
student_grade <- 87
cat("Grade of ", student_grade, ": ", classify_grade(student_grade), "\n")

student_grade <- 95
cat("Grade of ", student_grade, ": ", classify_grade(student_grade), "\n")

student_grade <- 72
cat("Grade of ", student_grade, ": ", classify_grade(student_grade), "\n")

```

11. APPLY THE RULE TO MULTIPLE **STUDENTS**

```r


# Define the determine_risk function
determine_risk <- function(attendance, grade) {
  if (attendance < 75 && grade < 75) {
    return("HIGH RISK")
  } else {
    return("LOW RISK")
  }
}

# Apply the function to the student data
student_df$RISK_LEVEL <- mapply(
  determine_risk,
  student_df$ATTENDANCE,
  student_df$GRADE
)

# Display the updated data
print(student_df)
```

12

```r
# Determine how many students belong to each risk category
risk_counts <- table(student_df$RISK_LEVEL)

print("Student Risk Category Counts:")
print(risk_counts)

# 1. HOW MANY STUDENTS ARE HIGH RISK?
high_risk_students <- ifelse("HIGH RISK" %in% names(risk_counts),
                             risk_counts["HIGH RISK"], 0)
cat("1. Number of students who are HIGH RISK:", high_risk_students, "\n")

# 2. HOW MANY ARE MODERATE RISK?
moderate_risk_students <- ifelse("MODERATE RISK" %in% names(risk_counts),
                                 risk_counts["MODERATE RISK"], 0)
cat("2. Number of students who are MODERATE RISK:", moderate_risk_students, "\n")

# 3. HOW MANY ARE LOW RISK?
low_risk_students <- ifelse("LOW RISK" %in% names(risk_counts),
                            risk_counts["LOW RISK"], 0)
cat("3. Number of students who are LOW RISK:", low_risk_students, "\n")

# 4. WHICH GROUP HAS THE LARGEST NUMBER OF STUDENTS?
if (length(risk_counts) > 0) {
  largest_group <- names(risk_counts)[which.max(risk_counts)]
  largest_count <- max(risk_counts)

  cat("4. The group with the largest number of students is '",
      largest_group, "' with ", largest_count, " students.\n", sep = "")
} else {
  print("4. No risk groups found in the data.")
}
```

```r

# 1. Extend the determine_risk function to include MODERATE RISK
determine_risk_extended <- function(attendance, grade) {

  if (attendance < 75 && grade < 75) {
    return("HIGH RISK")

  } else if ((attendance < 75 && grade >= 75) ||
             (attendance >= 75 && grade < 75)) {
    return("MODERATE RISK")

  } else {
    return("LOW RISK")
  }
}

# 2. Define a function to determine recommendation
determine_recommendation <- function(risk_level) {

  if (risk_level == "HIGH RISK") {
    return("IMMEDIATE ACADEMIC INTERVENTION")

  } else if (risk_level == "MODERATE RISK") {
    return("ACADEMIC MONITORING")

  } else if (risk_level == "LOW RISK") {
    return("CONTINUE REGULAR MONITORING")

  } else {
    return("UNKNOWN")
  }
}

# 3. Apply the extended risk determination
student_df$RISK_LEVEL <- mapply(
  determine_risk_extended,
  student_df$ATTENDANCE,
  student_df$GRADE
)

# 4. Apply the recommendation
student_df$RECOMMENDATION <- sapply(
  student_df$RISK_LEVEL,
  determine_recommendation
)

# Display the updated DataFrame
print(student_df)
```
